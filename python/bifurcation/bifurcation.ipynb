# Logistic Bifurcation and Route to Chaos

Iterated nonlinear maps can transition from fixed points to periodic orbits and chaos.
The logistic family
$$
x_{n+1}=r\,x_n(1-x_n),\qquad x_n\in[0,1],
$$
is a canonical model for this transition.

We build a bifurcation diagram and explore dependence on initial conditions and parameter $r$.


## Environment

We iterate maps on dense parameter grids and visualize asymptotic occupancy.


```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider

plt.rcParams["figure.dpi"] = 120
rng = np.random.default_rng(1)
```

## Logistic map iteration

Discard transient iterations and retain long-run states.


```python
def logistic_bifurcation(r_min=2.4, r_max=4.0, m=1600, n_trans=800, n_keep=260):
    r = np.linspace(r_min, r_max, m)
    x = rng.random(m)
    for _ in range(n_trans):
        x = r * x * (1 - x)
    cloud_x = []
    cloud_r = []
    for _ in range(n_keep):
        x = r * x * (1 - x)
        cloud_x.append(x.copy())
        cloud_r.append(r.copy())
    return np.concatenate(cloud_r), np.concatenate(cloud_x)

R, X = logistic_bifurcation()
```

## Bifurcation diagram

Plot asymptotic states versus parameter values.


```python
fig, ax = plt.subplots(figsize=(8.5, 4.6))
ax.plot(R, X, ",", alpha=0.35, color="black")
ax.set_xlim(2.4, 4.0)
ax.set_ylim(0, 1)
ax.set_title("Logistic bifurcation diagram")
ax.set_xlabel("r")
ax.set_ylabel("Asymptotic states")
plt.show()
```

## Orbit behavior at fixed parameter

Time-series and cobweb-like portrait for selected $r$.


```python
def orbit_plot(r=3.7):
    n = 240
    x = np.empty(n)
    x[0] = 0.231
    for i in range(n - 1):
        x[i + 1] = r * x[i] * (1 - x[i])

    fig, axes = plt.subplots(1, 2, figsize=(10, 3.6), constrained_layout=True)
    axes[0].plot(x[:120], lw=1.5)
    axes[0].set_title(f"Time series at r={r:.3f}")
    axes[0].grid(alpha=0.25)

    u = np.linspace(0, 1, 600)
    axes[1].plot(u, r * u * (1 - u), "b", lw=2, label="f(x)")
    axes[1].plot(u, u, "k--", lw=1, label="y=x")
    z = x[80]
    for _ in range(35):
        z1 = r * z * (1 - z)
        axes[1].plot([z, z], [z, z1], "r-", lw=0.8)
        axes[1].plot([z, z1], [z1, z1], "r-", lw=0.8)
        z = z1
    axes[1].set_xlim(0, 1)
    axes[1].set_ylim(0, 1)
    axes[1].set_aspect("equal")
    axes[1].set_title("Cobweb plot")
    axes[1].legend(loc="upper left")
    plt.show()

interact(orbit_plot, r=FloatSlider(min=2.5, max=4.0, step=0.002, value=3.7));
```
